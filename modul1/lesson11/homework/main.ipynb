Q14. Random Password Generator

Q25. Create a Simple Thread
Puzzle:
Create a simple function called say_hello() and run it using a thread.

```python
import threading

def say_hello():
    print("Hello!")

thread = threading.Thread(target=say_hello)

thread.start()
thread.join()
```
