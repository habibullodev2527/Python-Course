Q19. Wallet Class
Puzzle:
Create a class called Wallet with owner and money. Add methods to add money and spend money.

```python
class Wallet:
    def __init__(self, owner, money):
        self.owner = owner
        self.money = money

    def add(self, amount):
        self.money += amount

    def spend(self, amount):
        if amount <= self.money:
            self.money -= amount
        else:
            print("Not enough money.")

    def show_balance(self):
        print(f"{self.owner}: {self.money}")


wallet = Wallet("Ali", 100)

wallet.add(50)
wallet.spend(30)
wallet.show_balance()
```

Q21. Expense Tracker with Category Budgets & File Persistence
Puzzle:
Create a class ExpenseTracker that manages expenses and budgets.
Attributes:
•	expenses: a list of dictionaries, each with "category" (str) and "amount" (float/int).
•	budgets: a dictionary mapping category names to budget limits (float/int).
Methods:
1.	load_from_file(filename) – reads a text file where each line is category,amount. If the file does not exist, simply start with an empty expenses list. Handle ValueError for invalid numbers; skip the line and print a warning.
2.	add_expense(category, amount) – adds the expense to the list. If amount is not a positive number, raise ValueError.
3.	total_by_category(category) – returns the total amount spent in that category.
4.	set_budget(category, limit) – stores the budget limit for that category.
5.	budget_status() – returns a dictionary where keys are categories that have a budget set, and values are the remaining budget (budget – total spent). If the remaining amount is negative, include the string "OVERSHOOT" instead of the number.
6.	save_to_file(filename) – writes all expenses back to the file in the same category,amount format.
Example usage:
python
tracker = ExpenseTracker()
tracker.load_from_file("expenses.txt")
try:
    tracker.add_expense("food", 25000)
except ValueError as e:
    print(e)
tracker.set_budget("food", 50000)
tracker.set_budget("transport", 10000)
print(tracker.budget_status())  # food: 25000, transport: OVERSHOOT (if transport spending >10000)
tracker.save_to_file("expenses.txt")

```python
class ExpenseTracker:
    def __init__(self):
        self.expenses = []
        self.budgets = {}

    def load_from_file(self, filename):
        self.expenses = []

        try:
            with open(filename, "r") as file:
                for line in file:
                    line = line.strip()

                    if not line:
                        continue

                    try:
                        category, amount = line.split(",")
                        amount = float(amount)

                        self.expenses.append({
                            "category": category,
                            "amount": amount
                        })

                    except ValueError:
                        print(f"Warning: Invalid line skipped -> {line}")

        except FileNotFoundError:
            self.expenses = []

    def add_expense(self, category, amount):
        if amount <= 0:
            raise ValueError("Amount must be a positive number.")

        self.expenses.append({
            "category": category,
            "amount": amount
        })

    def total_by_category(self, category):
        total = 0

        for expense in self.expenses:
            if expense["category"] == category:
                total += expense["amount"]

        return total

    def set_budget(self, category, limit):
        self.budgets[category] = limit

    def budget_status(self):
        status = {}

        for category, budget in self.budgets.items():
            spent = self.total_by_category(category)
            remaining = budget - spent

            if remaining >= 0:
                status[category] = remaining
            else:
                status[category] = "OVERSHOOT"

        return status

    def save_to_file(self, filename):
        with open(filename, "w") as file:
            for expense in self.expenses:
                file.write(
                    f"{expense['category']},{expense['amount']}\n"
                )


tracker = ExpenseTracker()

tracker.load_from_file("expenses.txt")

try:
    tracker.add_expense("food", 25000)
    tracker.add_expense("transport", 12000)
except ValueError as e:
    print(e)

tracker.set_budget("food", 50000)
tracker.set_budget("transport", 10000)

print(tracker.total_by_category("food"))
print(tracker.budget_status())

tracker.save_to_file("expenses.txt")
```

Q22. Sales Report from File (Functions, No Classes)
Puzzle:
Write a function generate_sales_report(input_file, output_file) that reads a text file where each line represents one sale:
text
product_name,price,quantity_sold
Requirements:
1.	Reading the file:
o	Use try-except to handle FileNotFoundError (print an error and return None).
o	Skip any line that doesn’t have exactly three parts or where price / quantity_sold cannot be converted to float/int. Print a warning with the line number that is skipped.
2.	Calculations:
o	Build a dictionary revenue where keys are product names and values are the total revenue for that product (price * quantity_sold).
o	If the same product appears multiple times, add the revenue.
3.	Finding best sellers:
o	Determine the highest revenue value. Find all products that have that value (handle ties).
4.	Writing the output file:
o	Write the product list in alphabetical order, one per line: product: revenue.
o	After the list, add a blank line and then BEST SELLER(S): followed by the product name(s) and the revenue amount.
o	If the input file was empty or had no valid lines, write "No data available."
Example input (sales.txt):
text
Laptop,8000,2
Mouse,500,10
Laptop,8000,1
Keyboard,1200,5
Example output (report.txt):
text
Keyboard: 6000
Laptop: 24000
Mouse: 5000

BEST SELLER(S): Laptop (24000)
Function signature:
python
def generate_sales_report(input_file, output_file):
    # your code here
    # returns the revenue dictionary or None on error

```python
def generate_sales_report(input_file, output_file):
    revenue = {}

    try:
        with open(input_file, "r") as file:
            lines = file.readlines()
    except FileNotFoundError:
        print("Error: Input file not found.")
        return None

    for line_number, line in enumerate(lines, start=1):
        line = line.strip()

        if not line:
            continue

        parts = line.split(",")

        if len(parts) != 3:
            print(f"Warning: Line {line_number} skipped.")
            continue

        product = parts[0]

        try:
            price = float(parts[1])
            quantity = int(parts[2])
        except ValueError:
            print(f"Warning: Line {line_number} skipped.")
            continue

        total = price * quantity

        if product in revenue:
            revenue[product] += total
        else:
            revenue[product] = total

    with open(output_file, "w") as file:

        if not revenue:
            file.write("No data available.")
            return revenue

        for product in sorted(revenue):
            file.write(f"{product}: {revenue[product]}\n")

        highest = max(revenue.values())

        best_sellers = []

        for product, amount in revenue.items():
            if amount == highest:
                best_sellers.append(product)

        file.write("\n")

        file.write(
            f"BEST SELLER(S): {', '.join(best_sellers)} ({highest})"
        )

    return revenue


result = generate_sales_report("sales.txt", "report.txt")

print(result)
```

Q23. Text File Analyzer (Functions, No Classes)
Puzzle:
Write a function analyze_text(file_path) that processes a plain text file.
Requirements:
1.	File handling:
o	Use try-except to catch FileNotFoundError. Print an error and return None.
2.	Word extraction:
o	Read the whole file, convert to lowercase.
o	Split into words by spaces. Then, clean each word: remove punctuation from both ends of the word (you can check if the first/last character is not a letter using .isalpha() and slice accordingly).
o	Discard empty strings after cleaning.
3.	Statistics:
o	Count the total number of words (after cleaning).
o	Count the number of unique words.
o	Create a dictionary of word frequencies.
o	Find the most frequent word(s) and their count (handle ties).
o	Find the longest word(s) and their length (handle ties).
4.	Generating a report file:
o	Create a new file named "analysis_" + original_filename.
o	Write into it:
text
Total words: 
Unique words: 
Top 5 most frequent words:
   word1: frequency
   word2: frequency
   ...
Longest word(s) (length ):
   word1
   word2
o	If there are fewer than 5 unique words, show all of them sorted by frequency descending (and alphabetically for ties).
5.	Return value:
o	Return a tuple: (word_freq_dict, most_frequent_words, longest_words).
Example (input file contains "Hello world! Hello again. World, world."):
•	Words: hello, world, hello, again, world, world
•	Most frequent: world (3 times)
•	Longest: hello and again and world all length 5 (tie)
Output file "analysis_sample.txt" would contain the summary.

```python
import os


def analyze_text(file_path):
    try:
        with open(file_path, "r") as file:
            text = file.read().lower()
    except FileNotFoundError:
        print("Error: File not found.")
        return None

    words = []

    for word in text.split():

        while word and not word[0].isalpha():
            word = word[1:]

        while word and not word[-1].isalpha():
            word = word[:-1]

        if word:
            words.append(word)

    total_words = len(words)

    word_freq = {}

    for word in words:
        if word in word_freq:
            word_freq[word] += 1
        else:
            word_freq[word] = 1

    unique_words = len(word_freq)

    if word_freq:

        max_frequency = max(word_freq.values())

        most_frequent_words = []

        for word, count in word_freq.items():
            if count == max_frequency:
                most_frequent_words.append(word)

        max_length = max(len(word) for word in word_freq)

        longest_words = []

        for word in word_freq:
            if len(word) == max_length:
                longest_words.append(word)

    else:
        max_frequency = 0
        max_length = 0
        most_frequent_words = []
        longest_words = []

    sorted_words = sorted(
        word_freq.items(),
        key=lambda x: (-x[1], x[0])
    )

    output_file = "analysis_" + os.path.basename(file_path)

    with open(output_file, "w") as file:

        file.write(f"Total words: {total_words}\n")
        file.write(f"Unique words: {unique_words}\n")

        file.write("Top 5 most frequent words:\n")

        for word, count in sorted_words[:5]:
            file.write(f"   {word}: {count}\n")

        file.write(f"\nLongest word(s) (length {max_length}):\n")

        for word in longest_words:
            file.write(f"   {word}\n")

    return word_freq, most_frequent_words, longest_words


result = analyze_text("sample.txt")

print(result)
```

Q24. Inventory Selling System
Puzzle:
Create a class called Inventory.
The class should have a dictionary of products. Each product should have price and quantity.
Create methods:
1. add_product(name, price, quantity)
2. sell_product(name, quantity)
3. show_inventory()
When selling, decrease quantity only if enough stock exists.

```python
class Inventory:
    def __init__(self):
        self.products = {}

    def add_product(self, name, price, quantity):
        self.products[name] = {
            "price": price,
            "quantity": quantity
        }

    def sell_product(self, name, quantity):
        if name in self.products:
            if self.products[name]["quantity"] >= quantity:
                self.products[name]["quantity"] -= quantity
                print(f"{quantity} ta {name} sotildi.")
            else:
                print("Yetarli mahsulot mavjud emas.")
        else:
            print("Mahsulot topilmadi.")

    def show_inventory(self):
        for name, info in self.products.items():
            print(
                f"Product: {name}, "
                f"Price: {info['price']}, "
                f"Quantity: {info['quantity']}"
            )


inventory = Inventory()

inventory.add_product("Laptop", 8000, 10)
inventory.add_product("Mouse", 500, 20)

inventory.sell_product("Laptop", 3)
inventory.sell_product("Mouse", 5)

inventory.show_inventory()
```

Q25. Safe Student Score File Reader
Puzzle:
Create a file called scores.txt with these values:
Ali,80
Vali,abc
Madina,95
Sardor,70
Read the file. For each valid score, save it into a dictionary. If score is invalid, skip it and print an error message.

```python
scores = {}

with open("scores.txt", "r") as file:
    for line in file:
        name, score = line.strip().split(",")

        try:
            scores[name] = int(score)
        except ValueError:
            print(f"Invalid score for {name}. Skipping...")

print(scores)
```
