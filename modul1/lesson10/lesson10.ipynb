```python
password = input("Parolni kiriting: ")

upper = False
lower = False
digit = False

for i in password:
    if i.isupper():
        upper = True
    elif i.islower():
        lower = True
    elif i.isdigit():
        digit = True

if len(password) >= 8 and upper and lower and digit:
    print("Strong Password")
else:
    print("Weak Password")
```
