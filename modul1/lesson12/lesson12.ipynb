```python
names = ["Ali", "Jasur", "Muhammad", "Lola", "Aziza"]

def len_name(names):
    max_name = names[0]
    
    for i in names:
        if len(max_name) < len(i):
            max_name = i
    return max_name

print(len_name(names))
```

```python
def products(inventory: dict):
    count = 0
    for product, quantity in inventory.items():
        if quantity > 0:
            print(f"Available: {product}")
            count += 1
        else:
            print(f"Out of stock: {product}")

    return count


inventory = {
    "Keyboard": 12,
    "Mouse": 5,
    "Monitor": 0,
    "Laptop": 7,
    "Printer": 2
}

products(inventory)
```
