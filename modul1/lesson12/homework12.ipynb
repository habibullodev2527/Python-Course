Q16. Function to Count Negative Numbers
Puzzle:
Create a function called count_negative() that accepts a list of numbers and returns how many negative numbers are in the list.

```python
def count_negative(numbers):
    count = 0

    for number in numbers:
        if number < 0:
            count += 1

    return count


numbers = [5, -2, 10, -8, 0, -1, 7]

print(count_negative(numbers))
```

Q19. Create a Temperature Class
Puzzle:
Create a class called Temperature.
The class should have one attribute: celsius.
Create a method called to_fahrenheit() that converts Celsius to Fahrenheit.
Formula: fahrenheit = celsius * 9 / 5 + 32

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def to_fahrenheit(self):
        fahrenheit = self.celsius * 9 / 5 + 32
        return fahrenheit


temp = Temperature(25)

print(temp.to_fahrenheit())
```

Q21. Employee Payroll File Report
Puzzle:
Create a list of employee dictionaries. Each employee should have:
name, salary, bonus_percent
Calculate final salary for each employee and save the report into payroll.txt.

```python
employees = [
    {"name": "Ali", "salary": 3000, "bonus_percent": 10},
    {"name": "Vali", "salary": 4500, "bonus_percent": 15},
    {"name": "Hasan", "salary": 5000, "bonus_percent": 20}
]

with open("payroll.txt", "w") as file:
    for employee in employees:
        bonus = employee["salary"] * employee["bonus_percent"] / 100
        final_salary = employee["salary"] + bonus

        file.write(
            f"Name: {employee['name']}, "
            f"Salary: {employee['salary']}, "
            f"Bonus: {employee['bonus_percent']}%, "
            f"Final Salary: {final_salary}\n"
        )

print("Payroll report saved successfully.")
```

Q22. Course Score Analyzer Function
Puzzle:
Create a function called analyze_scores().
It should accept a dictionary where student names are keys and score lists are values.
The function should return a dictionary with each student’s average score.

```python
def analyze_scores(scores):
    averages = {}

    for student, marks in scores.items():
        average = sum(marks) / len(marks)
        averages[student] = average

    return averages


students = {
    "Ali": [80, 90, 100],
    "Vali": [70, 75, 85],
    "Hasan": [90, 95, 85]
}

print(analyze_scores(students))
```

Q23. Library Book Manager Class
Puzzle:
Create a class called BookManager.
The class should have a list of books.
Create these methods:
1. add_book(title)
2. remove_book(title)
3. search_book(title)
4. show_books()
search_book() should print whether the book exists or not.

```python
class BookManager:
    def __init__(self, books: list):
        self.books = books
    
    def add_book(self, title):
        self.books.append(title)
        
    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
        else:
            print(f"{title} topilmadi.")
        
    def search_book(self, title):
        if title in self.books:
            print(f"{title} - kitoblar ro'yxatida mavjud!")
        else:
            print(f"{title} - kitoblar ro'yxatida mavjud emas!")
            
    def show_books(self):
        for book in self.books:
            print(book)
    
books = [
    "Python Basics",
    "Clean Code",
    "Atomic Habits",
    "The Pragmatic Programmer",
    "Deep Learning"
]

book1 = BookManager(books)

book1.add_book("SQL")

book1.remove_book("Clean Code")

book1.search_book("SQL")

book1.show_books()
```
