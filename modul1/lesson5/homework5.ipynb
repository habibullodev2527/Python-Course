## Hard Questions

Q21. Progressive Income Tax Calculator
Puzzle:
Ask the user to enter their annual income (a number).
Calculate the tax according to progressive brackets:
•	Income up to 50,000: tax = 0%
•	Income from 50,001 to 100,000: tax = 10% on the amount above 50,000
•	Income above 100,000: tax = 5,000 (10% of the 50,000 bracket) + 20% on the amount above 100,000
Print the tax amount and the net income (income – tax).
Example:
Enter annual income: 120000
Tax: 9000
Net income: 111000

```python
income = float(input("Enter annual income: "))

if income <= 50000:
    tax = 0
elif income <= 100000:
    tax = (income - 50000) * 0.10
else:
    tax = 5000 + (income - 100000) * 0.20

net_income = income - tax

print("Tax:", tax)
print("Net income:", net_income)
```

Q22. Smart Ticket Pricing
Puzzle:
Ask the user to enter their age and the day of the week (e.g., "Monday", "Saturday").
Base prices:
•	Child (age < 12): 5
•	Adult (12 ≤ age < 60): 10
•	Senior (age ≥ 60): 7
If the day is Saturday or Sunday (case insensitive), add a weekend surcharge of 2.
Print the final ticket price.
Example:
Age: 65
Day: Sunday
Ticket price: 9

```python
age = int(input("Age: "))
day = input("Day: ").lower()

if age < 12:
    price = 5
elif age < 60:
    price = 10
else:
    price = 7

if day == "saturday" or day == "sunday":
    price += 2

print("Ticket price:", price)
```

Q23. Password Strength Checker
Puzzle:
Ask the user to enter a password (a string).
Evaluate its strength using these rules, all based on the characters inside it:
1.	Length is 8 or more → grants 1 point
2.	Contains at least one digit (0–9) → grants 1 point
3.	Contains at least one uppercase letter (A–Z) → grants 1 point
After checking, assign a strength label:
•	3 points → "Strong"
•	2 points → "Medium"
•	0 or 1 point → "Weak"
Print the password strength.
Hint: You can loop through the password characters and use .isdigit() and .isupper() to test them.
Example:
Enter password: Hello123
Strength: Strong
Enter password: abc
Strength: Weak


```python
password = input("Enter password: ")

points = 0
has_digit = False
has_upper = False

if len(password) >= 8:
    points += 1

for char in password:
    if char.isdigit():
        has_digit = True
    if char.isupper():
        has_upper = True

if has_digit:
    points += 1

if has_upper:
    points += 1

if points == 3:
    print("Strength: Strong")
elif points == 2:
    print("Strength: Medium")
else:
    print("Strength: Weak")
```
