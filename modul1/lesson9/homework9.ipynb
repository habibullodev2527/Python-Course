Q14. Student Average Method
Puzzle:
Create a class called Student with name and a list of scores. Add a method called calculate_average().

1-usul

```python
class Student :
    def __init__(self, name, scores: list):
        self.name = name
        self.scores = scores
        
    def calculate_average(self):
        summa = 0
        count = 0
        for score in self.scores:
            summa += score
            count += 1
        
        return summa / count
    
student1 = Student('Nizomjon', [90, 80, 70])
print(student1.calculate_average())
```

2-usul

```python
class Student:
    def __init__(self, name, scores: list):
        self.name = name
        self.scores = scores

    def calculate_average(self):
        return sum(self.scores) / len(self.scores)


student1 = Student("Nizomjon", [90, 80, 70])
print(student1.calculate_average())
```

Q21. Shopping Cart Class
Puzzle:
Create a class called ShoppingCart.

The class should have:
1. Customer name
2. Empty list called items

Create these methods:
1. add_item(item_name) - adds item to list
2. remove_item(item_name) - removes item from list
3. show_cart() - prints customer name and cart items

```python
class ShoppingCart:
    def __init__(self, name):
        self.name = name
        self.items = []
        
    def add_item(self, item_name):
        self.items.append(item_name)
    
    def remove_item(self, item_name):
        if item_name in self.items:
            self.items.remove(item_name)
        else:
            print("Mahsulot savatda mavjud emas.")
    
    def show_cart(self):
        print(f"Mijoz nomi: {self.name}\nSavatdagi mahsulotlar ro'yxati: {self.items}")
        
cart1 = ShoppingCart('Ali')

cart1.add_item('apple')
cart1.add_item('bread')
cart1.add_item('banana')

cart1.remove_item('bread')

cart1.show_cart()
```

Q22. Student Management Class
Puzzle:
Create a class called Student.

The class should have:
1. Student name
2. List of scores

Create these methods:
1. add_score(score) - adds score to the list
2. calculate_average() - returns average score
3. show_result() - prints "Pass" if average is 60 or higher, otherwise "Fail"

```python
class Student:
    def __init__(self, student_name, scores: list):
        self.student_name = student_name
        self.scores = scores
        
    def add_score(self, score):
        self.scores.append(score)
        
    def calculate_average(self):
        return sum(self.scores) / len(self.scores)
    
    def show_result(self):
        if self.calculate_average() >= 60:
            return "Pass"
        else:
            return "Fail"
            
student1 = Student('Vali', [90, 60, 50])

student1.add_score(40)

print(f"O'rtacha ball: {student1.calculate_average()}")

print(student1.show_result())
```

Q23. Simple Bank Account System
Puzzle:
Create a class called BankAccount.

The class should have:
1. Owner name
2. Balance

Create these methods:
1. deposit(amount) - adds money to balance
2. withdraw(amount) - subtracts money from balance if enough money exists
3. show_balance() - prints owner and balance

```python
class BankAccount:
    def __init__(self, owner_name, balance):
        self.owner_name = owner_name
        self.balance = balance
        
    def deposit(self, amount):
        self.balance += amount
        
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else: 
            print("Balansingizda mablag' yetarli emas!")
            
    def show_balance(self):
        print(f"Hisob raqam egasi: {self.owner_name}, hisobdagi balans: ${self.balance}")
        
bankaccount1 = BankAccount('Umaraliyev Nizomjon', 500)

bankaccount1.deposit(100)

bankaccount1.withdraw(50)

bankaccount1.show_balance()
```
