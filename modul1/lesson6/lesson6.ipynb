Q22. Word Frequency Counter (Dictionary + Max Search)
Puzzle:
Ask the user to enter a sentence (a string).
1.	Split the sentence into words (ignore punctuation for simplicity – assume words are separated by spaces).
2.	Convert all words to lowercase so that “Python” and “python” count as the same word.
3.	Count how many times each word appears using a dictionary.
4.	Print each word and its frequency in alphabetical order (sorted by word).
5.	Find and print the most frequent word(s) and their count.
o	If there is a tie, print all words that share the highest frequency.
Example:
text
Enter a sentence: The cat and the dog and the mouse
Word frequencies:
and : 2
cat : 1
dog : 1
mouse : 1
the : 3
Most frequent: the (3)

```python
sentence = input("Gap kiriting: ").lower()
words = sentence.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print("Word frequencies:")
for word in sorted(freq):
    print(word, ":", freq[word])

max_count = 0
for word in freq:
    if freq[word] > max_count:
        max_count = freq[word]

most_words = []
for word in freq:
    if freq[word] == max_count:
        most_words.append(word)

print("Most frequent:", ", ".join(most_words), f"({max_count})")
```

Q23. Matrix Row and Column Sums (Nested Loops)
Puzzle:
You are given a 2D list (a list of lists) representing a matrix of numbers. For example:
python
matrix = [
    [2, 5, 3],
    [7, 1, 8],
    [4, 6, 2]
]
Write a program that uses nested loops to:
1.	Print the matrix row by row.
2.	Calculate and print the sum of each row.
3.	Calculate and print the sum of each column.
Do not use any built in sum function; accumulate manually with loops.
Example output:
text
Matrix:
2 5 3
7 1 8
4 6 2

Row sums:
Row 1 sum: 10
Row 2 sum: 16
Row 3 sum: 12

Column sums:
Column 1 sum: 13
Column 2 sum: 12
Column 3 sum: 13
Bonus (optional harder): Make the matrix size dynamic by asking the user for the number of rows and columns, then let them enter each element one by one.

```python
rows = int(input("Qatorlar sonini kiriting: "))
cols = int(input("Ustunlar sonini kiriting: "))

matrix = []

for i in range(rows):
    row = []
    for j in range(cols):
        num = int(input(f"{i+1}-qator {j+1}-ustun elementini kiriting: "))
        row.append(num)
    matrix.append(row)

print("\nMatrix:")
for row in matrix:
    for num in row:
        print(num, end=" ")
    print()

print("\nRow sums:")
for i in range(len(matrix)):
    row_sum = 0
    for j in range(len(matrix[i])):
        row_sum += matrix[i][j]
    print(f"Row {i+1} sum: {row_sum}")

print("\nColumn sums:")
for j in range(len(matrix[0])):
    col_sum = 0
    for i in range(len(matrix)):
        col_sum += matrix[i][j]
    print(f"Column {j+1} sum: {col_sum}")
```
