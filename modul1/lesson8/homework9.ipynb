Q11. Raise ValueError for Invalid Integer
Puzzle:
Ask the user to enter an integer. If the input is not a valid integer, raise and handle a ValueError.

```python
try:
    number = input("Butun son kiriting: ")

    if not number.lstrip("-").isdigit():
        raise ValueError("Butun son kiritilmadi!")

    number = int(number)
    print(number)

except ValueError as e:
    print(e)
```

Q12. Raise TypeError for Non-Numeric Inputs
Puzzle:
Ask the user to enter two values. If the values are not numerical, raise and handle a TypeError.

```python
try:
    num1 = input("Birinchi qiymatni kiriting: ")
    num2 = input("Ikkinchi qiymatni kiriting: ")

    if not num1.replace(".", "", 1).isdigit() or not num2.replace(".", "", 1).isdigit():
        raise TypeError("Faqat son kiriting!")

    num1 = float(num1)
    num2 = float(num2)

    print(num1, num2)

except TypeError as e:
    print(e)
```

Q20. Safe File Reader Function
Puzzle:
Create a function called read_file() that accepts a file name. The function should read and print file content. Handle FileNotFoundError.

```python
def read_file(filename):
    try:
        with open(filename, "r") as file:
            print(file.read())
    except FileNotFoundError:
        print("Bunday fayl mavjud emas!")

read_file("students.txt")
```

## Hard Questions

Q21. Student Grade File System
Puzzle:
Ask the user to enter student name and 3 exam scores. Handle wrong score input.
If input is correct:
•	Calculate total score
•	Calculate average score
•	Save name, total, and average into student_report.txt

```python
total = 0
count = 0

try:
    student_name = input('Talabani ismini kiriting: ')
    
    for i in range(3):
        score = int(input('Talabani imtihon ballini kiriting: '))
        total += score
        count += 1 
        
    avg = total / count
    
    with open('student_report.txt', 'w') as file:
        file.write(f"Name: {student_name}\n")
        file.write(f"Umumiy ball: {total}\n")
        file.write(f"O'rtacha ball: {avg}")
    
except ValueError:
    print("Noto'g'ri ball kiritdingiz!")
```

Q23. Read Numbers From File and Calculate Total
Puzzle:
Create a file called numbers.txt with these values:
10
20
abc
30
Read the file and calculate the total of valid numbers only. If a line is not a number, skip it and print a message.

```python
total = 0

with open("numbers.txt", "r") as file:
    for item in file:
        try:
            total += int(item)
        except ValueError:
            print(f"{item.strip()} son emas, o'tkazib yuborildi.")

print("Jami:", total)
```
