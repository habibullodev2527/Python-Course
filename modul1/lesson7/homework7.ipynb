Q16. Digit Sum
Puzzle:
Write a function called digit_sum(k) that calculates the sum of the digits of the number k.

1-usul

```python
def digit_sum(k):
    return sum(int(digit) for digit in str(k))

k = int(input("Son kiriting: "))
print(digit_sum(k))
```

2-usul

```python
def digit_sum(k):
    total = 0
    while k > 0:
        total += k % 10
        k //= 10
    return total

k = int(input("Son kiriting: "))
print(digit_sum(k))
```

Q18. Count Even Numbers in List
Puzzle:
Create a function called count_even_numbers() that accepts a list of numbers and returns how many even numbers are in the list.

1-usul

```python
def count_even_numbers(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count

numbers = [1, 2, 3, 4, 5, 6, 8]
print(count_even_numbers(numbers))
```

2-usul

```python
def count_even_numbers(numbers):
    return sum(1 for number in numbers if number % 2 == 0)

numbers = [1, 2, 3, 4, 5, 6, 8]
print(count_even_numbers(numbers))
```

Q19. Check Prime Number
Puzzle:
Define a function is_prime(n) which returns True if the given n (n > 0) is a prime number, otherwise returns False.

Hint: a number should be > 0, not equal to 1, and should divide only by itself and 1.

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


number = int(input("Son kiriting: "))
print(is_prime(number))
```

Q21. Grade Analyzer with Custom Passing Threshold
Puzzle:
Write a function analyze_grades(name, scores, passing_score=60) that:
•	Accepts a student’s name (string) and a list of scores (any number of integers/floats).
•	Accepts an optional passing_score parameter with a default value of 60.
•	Calculates the total, average, highest, and lowest score.
•	Determines a letter grade based on the average:
o	A: 90–100
o	B: 80–89
o	C: 70–79
o	D: 60–69
o	F: below 60
•	Determines the status: "Pass" if the average is greater than or equal to passing_score, otherwise "Fail".
•	Returns a dictionary with all the computed information.
Example call:
python
result = analyze_grades("Ali", [85, 92, 78], passing_score=70)
Expected return:
python
{
    "name": "Ali",
    "total": 255,
    "average": 85.0,
    "highest": 92,
    "lowest": 78,
    "letter_grade": "B",
    "status": "Pass"
}

```python
def analyze_grades(name, scores, passing_score=60):
    total = sum(scores)
    average = total / len(scores)
    highest = max(scores)
    lowest = min(scores)

    if average >= 90:
        letter_grade = "A"
    elif average >= 80:
        letter_grade = "B"
    elif average >= 70:
        letter_grade = "C"
    elif average >= 60:
        letter_grade = "D"
    else:
        letter_grade = "F"

    if average >= passing_score:
        status = "Pass"
    else:
        status = "Fail"

    return {
        "name": name,
        "total": total,
        "average": average,
        "highest": highest,
        "lowest": lowest,
        "letter_grade": letter_grade,
        "status": status
    }


result = analyze_grades("Ali", [85, 92, 78], passing_score=70)
print(result)
```

Q22. Shopping Cart with Tiered Discounts
Puzzle:
Write a function total_cart_with_discount(cart, discount_rate=0.0) that:
•	Accepts a list of dictionaries cart. Each dictionary represents an item and has the keys "name", "price" (float/int), and "quantity" (int).
•	Accepts an optional discount_rate (a float, e.g. 0.1 for 10%), default 0.0.
•	Calculates the subtotal by summing price * quantity for all items.
•	Applies an extra automatic discount of 5% (0.05) if the subtotal is strictly greater than 100,000. The final discount rate becomes discount_rate + 0.05. Otherwise, it stays discount_rate.
•	Computes the discount amount and final total after discount.
•	Returns a tuple: (subtotal, discount_amount, final_total).
•	The function must not print anything.
Example call:
python
cart = [
    {"name": "Laptop", "price": 80000, "quantity": 2},
    {"name": "Mouse",  "price": 5000,  "quantity": 3}
]
subtotal, disc, total = total_cart_with_discount(cart, discount_rate=0.1)
Expected result:
Subtotal = 175000 (>100000), so final discount = 0.1 + 0.05 = 0.15
discount_amount = 26250, final_total = 148750.

```python
def total_cart_with_discount(cart, discount_rate=0.0):
    subtotal = 0

    for item in cart:
        subtotal += item["price"] * item["quantity"]

    if subtotal > 100000:
        discount_rate += 0.05

    discount_amount = subtotal * discount_rate
    final_total = subtotal - discount_amount

    return subtotal, discount_amount, final_total


cart = [
    {"name": "Laptop", "price": 80000, "quantity": 2},
    {"name": "Mouse", "price": 5000, "quantity": 3}
]

subtotal, disc, total = total_cart_with_discount(cart, discount_rate=0.1)

print("Subtotal:", subtotal)
print("Discount:", disc)
print("Final Total:", total)
```

Q23. First Unique Product
Puzzle:
Write a function called first_unique(items) that accepts a list of product names. The function should return the first product that appears only once in the list. If every product appears more than once, return "No unique product".

Example 1:
items = ["apple", "banana", "apple", "orange", "banana", "kiwi"]

Result:
"orange"

Explanation:
apple  → appears 2 times
banana → appears 2 times
orange → appears 1 time
kiwi   → appears 1 time
The answer is "orange", not "kiwi", because "orange" is the first unique product in the original list.

Example 2:
items = ["apple", "banana", "apple", "banana"]

Result:
"No unique product"

```python
def first_unique(items):
    for item in items:
        if items.count(item) == 1:
            return item

    return "No unique product"


items = ["apple", "banana", "apple", "orange", "banana", "kiwi"]
print(first_unique(items))
```
