Q21. Student Profile With Nested Dictionary
Puzzle:
Create a student dictionary with:
1. Name
2. Age
3. Nested dictionary called grades with math, English, and Python grades
4. List called subjects

Do these tasks:
1. Change Python grade to 100
2. Add "SQL" to the subjects list
3. Calculate average grade
4. Print the updated student dictionary
5. Print the average grade

```python
student = {
    "name": "Ali",
    "age": 20,
    "grades": {
        "math": 80,
        "english": 85,
        "python": 95
    },
    "subjects": ["Math", "English", "Python", "Physics"]
}

student["grades"]["python"] = 100
student["subjects"].append("SQL")
avg_grade = (student["grades"]["math"] + student["grades"]["english"] + student["grades"]["python"]) / 3

print(student)
print(avg_grade)
```

Q22. Product Stock System With Nested Dictionary
Puzzle:
Create a product dictionary with:
1. Product name
2. Nested dictionary called details with price and quantity
3. List called colors

```python
product = {
    "name": "Laptop",
    "details": {
        "price": 1200,
        "quantity": 10
    },
    "colors": ["Silver", "White"]
}

product["details"]["quantity"] += 5
product["details"]["price"] *= 1.10
product["colors"].append("Black")
total_stock = product["details"]["price"] * product["details"]["quantity"]

print(product)
print("Total stock value:", total_stock)
```

Q23. Course Students With Sets
Puzzle:
Create two sets:

python_students = {"Ali", "Vali", "Madina", "Sardor"}
sql_students = {"Madina", "Sardor", "Aziza", "Jasur"}

Print:
1. All unique students
2. Students studying both Python and SQL
3. Students studying only Python
4. Students studying only SQL

Easy Questions From Next Theme - Conditional Statements

```python
python_students = {"Ali", "Vali", "Madina", "Sardor"}
sql_students = {"Madina", "Sardor", "Aziza", "Jasur"}

print(python_students | sql_students)
print(python_students & sql_students)
print(python_students - sql_students)
print(sql_students - python_students)
```
