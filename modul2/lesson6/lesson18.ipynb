Vazifa sharti:

Sizga do'konning savdo ma'lumotlari lug'at (dictionary) shaklida berilgan. Siz ushbu ma'lumotlardan foydalanib quyidagi amallarni bajarishingiz kerak:

DataFrame yaratish: Berilgan data lug'atidan df nomli DataFrame yarating.

Ustunlarni qayta nomlash: prod_name ustunini product, old_price ustunini price deb o'zgartiring.

Tozalash va turlarni o'zgartirish: price ustunidagi barcha bo'sh (NaN) qiymatlarni 0 bilan to'ldiring va turini int (butun son) ga 

o'tkazing.

Yangi ustun qo'shish: is_expensive nomli yangi ustun yarating. Agar mahsulot narxi 500 dan katta bo'lsa True, aks holda False bo'lsin.

Mahsulotlar sonini (size), o'rtacha narxni va unikal mahsulotlar nomini aniqlang.

O'chirish: rating ustunini DataFrame'dan butunlay o'chirib tashlang.

Statistika: Faqatgina price va is_expensive ustunlari uchun umumiy statistikani (describe()) chiqaring.

```python
import pandas as pd

data = {
    'prod_name': ['iPhone 15', 'Samsung S23', 'MacBook Air', 'AirPods', 'Apple Watch', 'Mouse'],
    'old_price': [1200, 900, 1500, 200, None, 50],
    'rating': [4.8, 4.5, 4.9, 4.2, 4.7, 3.5],
    'stock': [10, 15, 7, 25, 12, 50]
}

# 1
df = pd.DataFrame(data)
df

# 2
df = df.rename(columns={'prod_name':'product', 'old_price':'price'})
df

# 3
df['price'] = df['price'].fillna(0).astype(int)
df

# 4
df['is_expensive'] = df['price'] > 500
df

# 5
df.size
print(df['price'].mean())
df['product'].unique()

# 6

df.drop(columns=['rating'], inplace=True)
df

# 7
df[['is_expensive', 'price']].describe()
```

```python
filter1 = movie.imdb_score > 8
filter2 = movie.content_rating == 'PG-13'
filter3 = (movie.title_year < 2000) | (movie.title_year > 2009)
```

```python
result = filter1 & filter2 & filter3
result
```

```python
movie[result]
```

1 - imdb_score > 7
2 - content_rating = "R"
3 - actor_3_facebook_likes > 500 and title_year > 2010

```python
filter_b = movie.imdb_score > 7
filter_b2 = movie.content_rating == 'R'
filter_b3 = (movie.actor_3_facebook_likes > 500) & (movie.title_year > 2010)

natija = filter_b & filter_b2 & filter_b3
natija
```

```python
movie[natija]
```

```python
all_filter = result | natija
all_filter
```

```python
movie[all_filter]
```

```python
cols = ['director_name', 'director_facebook_likes', 'actor_2_name', 'actor_2_facebook_likes']
movie_filtered = movie.loc[all_filter, cols]
movie_filtered
```

```python
depts = ['Houston Police Department-HPD', 'Houston Fire Department (HFD)']
filter_dept = employee.DEPARTMENT.isin(depts)
filter_gender = employee.GENDER == 'Female'
filter_sal = employee.BASE_SALARY.between(80000, 120000)
```

```python
all = filter_dept & filter_gender & filter_sal
all
```

```python
employee.loc[all, ['UNIQUE_ID', 'DEPARTMENT', 'EMPLOYMENT_TYPE', 'GENDER']]
```

* query() -> SQL where clause bilan bir xil ishlaydi

```python
depts = ['Houston Police Department-HPD', 'Houston Fire Department (HFD)']

cols = ['UNIQUE_ID', 'DEPARTMENT', 'EMPLOYMENT_TYPE', 'GENDER']

q = "DEPARTMENT in @depts and GENDER == 'Female' and 80000 <= BASE_SALARY <= 120000"
emp_filtered = employee.query(q)
```

```python
emp_filtered[cols]
```

```python
employee.query("BASE_SALARY > 70000")
```
