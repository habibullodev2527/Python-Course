```python
products = pd.read_csv('data/olist_products_dataset.csv')
products
```

```python
orders = pd.read_csv('data/olist_orders_dataset.csv')
orders
```

```python
order_items = pd.read_csv('data/olist_order_items_dataset.csv')

order_items
```

```python
product_category_name = pd.read_csv('data/product_category_name_translation.csv')
product_category_name
```

```python
result = (
    order_items
    .merge(products, on='product_id')
    .merge(product_category_name, on='product_category_name')
    .merge(orders, on='order_id')
)

cancelled = (
    result[result['order_status'] == 'canceled']
    .groupby('product_category_name_english', as_index=False)
    .agg(cancelled_orders=('order_status', 'count'))
    .sort_values('cancelled_orders', ascending=False)
)

cancelled
```

```python
import sqlite3

conn = sqlite3.connect('data/chinook.db')

artists = pd.read_sql("SELECT *FROM artists", conn)

albums = pd.read_sql("SELECT *FROM albums", conn)

full_version = (
    albums
    .merge(artists, on='ArtistId')
)

full_version['full_version'] = (
    full_version['Title'] + ' by ' + full_version['Name']
)

full_version[['full_version']]

conn.close()
```
