Q21. Profitability Intelligence Dashboard — Using enterprise_data.csv, create Profit, Profit Margin, and Efficiency Score columns using Monthly Sales, Monthly Expenses, and Project_Score. Identify top 5 employees by Efficiency Score, detect employees with high sales but low efficiency, rank departments by profitability, and export profitability_dashboard.csv.

```python
import pandas as pd
import numpy as np

# CSV faylni o'qish
enterprise_df = pd.read_csv("enterprise_data.csv")

# 1. Profit ustunini yaratish
enterprise_df["Profit"] = (
    enterprise_df["Monthly Sales"] -
    enterprise_df["Monthly Expenses"]
)

# 2. Profit Margin (%)
enterprise_df["Profit_Margin"] = (
    enterprise_df["Profit"] /
    enterprise_df["Monthly Sales"]
) * 100

# 3. Efficiency Score
enterprise_df["Efficiency_Score"] = (
    enterprise_df["Profit"] *
    enterprise_df["Project_Score"] /
    100
)

# -------------------------------
# Top 5 employees by Efficiency Score
# -------------------------------

top5 = (
    enterprise_df[
        [
            "First Name",
            "Last Name",
            "Department",
            "Efficiency_Score"
        ]
    ]
    .sort_values(by="Efficiency_Score", ascending=False)
    .head(5)
)

print("Top 5 Employees")
print(top5)

# -------------------------------
# High Sales but Low Efficiency
# -------------------------------

high_sales = enterprise_df["Monthly Sales"].median()
low_efficiency = enterprise_df["Efficiency_Score"].median()

low_eff = enterprise_df[
    (enterprise_df["Monthly Sales"] > high_sales) &
    (enterprise_df["Efficiency_Score"] < low_efficiency)
]

print("\nHigh Sales but Low Efficiency")
print(
    low_eff[
        [
            "First Name",
            "Last Name",
            "Department",
            "Monthly Sales",
            "Efficiency_Score"
        ]
    ]
)

# -------------------------------
# Department Profitability Ranking
# -------------------------------

department_profit = (
    enterprise_df
    .groupby("Department", as_index=False)
    .agg(
        Total_Profit=("Profit", "sum"),
        Avg_Profit_Margin=("Profit_Margin", "mean")
    )
    .sort_values(
        by="Total_Profit",
        ascending=False
    )
)

print("\nDepartment Profitability")
print(department_profit)

# -------------------------------
# CSV faylga eksport qilish
# -------------------------------

enterprise_df.to_csv(
    "profitability_dashboard.csv",
    index=False
)

print("\nprofitability_dashboard.csv yaratildi.")
```

Q23. - read the CSV file into a DataFrame
- print only First Name, Last Name, and Department columns
- print only Monthly Sales and Monthly Expenses columns
- create a new column called Full_Name using First Name and Last Name
- create a new column called Profit using Monthly Sales - Monthly Expenses
- print the first 10 rows of the updated DataFrame

```python
import pandas as pd

# CSV faylni o'qish
enterprise_df = pd.read_csv("enterprise_data.csv")

# 1. Faqat First Name, Last Name va Department ustunlarini chiqarish
print("First Name, Last Name va Department:")
print(enterprise_df[["First Name", "Last Name", "Department"]])

# 2. Faqat Monthly Sales va Monthly Expenses ustunlarini chiqarish
print("\nMonthly Sales va Monthly Expenses:")
print(enterprise_df[["Monthly Sales", "Monthly Expenses"]])

# 3. Full_Name ustunini yaratish
enterprise_df["Full_Name"] = (
    enterprise_df["First Name"] + " " + enterprise_df["Last Name"]
)

# 4. Profit ustunini yaratish
enterprise_df["Profit"] = (
    enterprise_df["Monthly Sales"] -
    enterprise_df["Monthly Expenses"]
)

# 5. Yangilangan DataFrame ning birinchi 10 qatorini chiqarish
print("\nUpdated DataFrame (First 10 Rows):")
print(enterprise_df.head(10))
```
