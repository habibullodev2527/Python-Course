```python
from datetime import datetime

login_times = [
    "2026-03-11 08:45:00",
    "2026-03-11 09:10:00",
    "2026-03-11 09:40:00",
    "2026-03-11 10:05:00"
]

for i in range(len(login_times) - 1):
    t1 = datetime.strptime(login_times[i], "%Y-%m-%d %H:%M:%S")
    t2 = datetime.strptime(login_times[i + 1], "%Y-%m-%d %H:%M:%S")

    difference = int((t2 - t1).total_seconds() // 60)

    print(f"Difference: {difference} minut")
```
