Q21. Weather Intelligence Analytics Dashboard — Using OpenWeatherMap API, fetch weather data for Tashkent, Tokyo, London, Dubai, and New York. Extract temperature, humidity, pressure, visibility, and wind speed. Calculate average temperature, detect cities with extreme humidity, classify cities into Hot/Moderate/Cold groups, rank cities by comfort score, and export weather_dashboard.json.

```python
import requests
import json

API_KEY = "65de0710bebf24c26272403486b4df8d"

cities = [
    "Tashkent",
    "Tokyo",
    "London",
    "Dubai",
    "New York"
]

weather_dashboard = []

for city in cities:

    url = (
        f"https://api.openweathermap.org/data/2.5/weather"
        f"?q={city}"
        f"&appid={API_KEY}"
        f"&units=metric"
    )

    response = requests.get(url)

    if response.status_code == 200:

        data = response.json()

        temperature = data["main"]["temp"]
        humidity = data["main"]["humidity"]
        pressure = data["main"]["pressure"]
        visibility = data.get("visibility", 0)
        wind_speed = data["wind"]["speed"]

        # Temperature Classification
        if temperature >= 30:
            category = "Hot"
        elif temperature >= 15:
            category = "Moderate"
        else:
            category = "Cold"

        # Comfort Score
        comfort_score = (
            100
            - abs(temperature - 22) * 2
            - abs(humidity - 50) * 0.5
            - wind_speed * 2
        )

        weather_dashboard.append({
            "city": city,
            "temperature": temperature,
            "humidity": humidity,
            "pressure": pressure,
            "visibility": visibility,
            "wind_speed": wind_speed,
            "category": category,
            "comfort_score": round(comfort_score, 2)
        })

    else:
        print(f"{city} ma'lumotini olishda xatolik!")

# ===========================
# Average Temperature
# ===========================

average_temperature = (
    sum(city["temperature"] for city in weather_dashboard)
    / len(weather_dashboard)
)

print("=" * 60)
print(f"Average Temperature : {average_temperature:.2f}°C")
print("=" * 60)

# ===========================
# Extreme Humidity
# ===========================

print("\nCities with Extreme Humidity (>=80%)")

found = False

for city in weather_dashboard:

    if city["humidity"] >= 80:
        print(f"{city['city']} ({city['humidity']}%)")
        found = True

if not found:
    print("No city found.")

# ===========================
# Temperature Category
# ===========================

print("\nTemperature Categories")

for city in weather_dashboard:

    print(
        f"{city['city']:<12}"
        f" -> {city['category']}"
    )

# ===========================
# Comfort Ranking
# ===========================

weather_dashboard.sort(
    key=lambda city: city["comfort_score"],
    reverse=True
)

print("\nComfort Score Ranking")

for index, city in enumerate(weather_dashboard, start=1):

    print(
        f"{index}. "
        f"{city['city']:<12}"
        f" Score: {city['comfort_score']}"
    )

# ===========================
# Highest Temperature
# ===========================

hottest = max(
    weather_dashboard,
    key=lambda city: city["temperature"]
)

print("\nHottest City")
print(
    hottest["city"],
    f"({hottest['temperature']}°C)"
)

# ===========================
# Lowest Temperature
# ===========================

coldest = min(
    weather_dashboard,
    key=lambda city: city["temperature"]
)

print("\nColdest City")
print(
    coldest["city"],
    f"({coldest['temperature']}°C)"
)

# ===========================
# Highest Humidity
# ===========================

highest_humidity = max(
    weather_dashboard,
    key=lambda city: city["humidity"]
)

print("\nHighest Humidity")

print(
    highest_humidity["city"],
    f"({highest_humidity['humidity']}%)"
)

# ===========================
# Highest Wind Speed
# ===========================

highest_wind = max(
    weather_dashboard,
    key=lambda city: city["wind_speed"]
)

print("\nHighest Wind Speed")

print(
    highest_wind["city"],
    f"({highest_wind['wind_speed']} m/s)"
)

# ===========================
# Save JSON
# ===========================

with open(
    "weather_dashboard.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        weather_dashboard,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\nweather_dashboard.json created successfully!")
```

Q22. Movie Recommendation & Scoring Engine — Using OMDb API, analyze at least 8 movies. Extract IMDb rating, runtime, genre, language, and box office values. Calculate recommendation scores, identify highest-rated genres, detect underrated movies, rank movies by recommendation score, and export movie_analytics.csv.

```python
import requests
import csv

API_KEY = "b5ee201"

movies = [
    "Inception",
    "Interstellar",
    "The Dark Knight",
    "Avengers: Endgame",
    "The Matrix",
    "Titanic",
    "Forrest Gump",
    "Parasite"
]

movie_data = []

for movie in movies:

    url = f"http://www.omdbapi.com/?apikey={API_KEY}&t={movie}"

    response = requests.get(url)

    if response.status_code == 200:

        data = response.json()

        if data["Response"] == "True":

            imdb = float(data["imdbRating"]) if data["imdbRating"] != "N/A" else 0

            runtime = (
                int(data["Runtime"].split()[0])
                if data["Runtime"] != "N/A"
                else 0
            )

            genres = data["Genre"]
            language = data["Language"]

            if data["BoxOffice"] != "N/A":
                box_office = int(
                    data["BoxOffice"]
                    .replace("$", "")
                    .replace(",", "")
                )
            else:
                box_office = 0

            recommendation_score = (
                imdb * 10
                + runtime / 10
                + box_office / 100000000
            )

            movie_data.append({
                "Title": data["Title"],
                "IMDb Rating": imdb,
                "Runtime": runtime,
                "Genre": genres,
                "Language": language,
                "BoxOffice": box_office,
                "Recommendation Score": round(recommendation_score, 2)
            })

        else:
            print(f"{movie} topilmadi!")

    else:
        print("API bilan bog'lanishda xatolik!")

# ======================================================
# Average IMDb Rating
# ======================================================

average_rating = (
    sum(movie["IMDb Rating"] for movie in movie_data)
    / len(movie_data)
)

print("=" * 60)
print(f"Average IMDb Rating : {average_rating:.2f}")
print("=" * 60)

# ======================================================
# Highest Rated Genres
# ======================================================

genre_scores = {}

for movie in movie_data:

    for genre in movie["Genre"].split(", "):

        if genre not in genre_scores:
            genre_scores[genre] = []

        genre_scores[genre].append(movie["IMDb Rating"])

print("\nAverage Rating by Genre")
print("-" * 40)

genre_average = {}

for genre, ratings in genre_scores.items():

    avg = sum(ratings) / len(ratings)

    genre_average[genre] = avg

    print(f"{genre:<15} {avg:.2f}")

best_genre = max(
    genre_average,
    key=genre_average.get
)

print("\nHighest Rated Genre")
print(best_genre, "-", round(genre_average[best_genre], 2))

# ======================================================
# Underrated Movies
# ======================================================

print("\nUnderrated Movies")
print("-" * 40)

found = False

for movie in movie_data:

    if (
        movie["IMDb Rating"] >= 8.5
        and movie["BoxOffice"] < 300000000
    ):

        print(
            movie["Title"],
            f"(IMDb: {movie['IMDb Rating']})"
        )

        found = True

if not found:
    print("No underrated movies found.")

# ======================================================
# Recommendation Ranking
# ======================================================

movie_data.sort(
    key=lambda movie: movie["Recommendation Score"],
    reverse=True
)

print("\nRecommendation Ranking")
print("-" * 40)

for index, movie in enumerate(movie_data, start=1):

    print(
        f"{index}. "
        f"{movie['Title']:<25}"
        f"{movie['Recommendation Score']}"
    )

# ======================================================
# Best Recommended Movie
# ======================================================

best_movie = movie_data[0]

print("\nBest Recommended Movie")
print("-" * 40)

print(best_movie["Title"])
print("IMDb:", best_movie["IMDb Rating"])
print("Recommendation Score:", best_movie["Recommendation Score"])

# ======================================================
# Highest Box Office
# ======================================================

highest_box = max(
    movie_data,
    key=lambda movie: movie["BoxOffice"]
)

print("\nHighest Box Office")
print("-" * 40)

print(
    highest_box["Title"],
    f"${highest_box['BoxOffice']:,}"
)

# ======================================================
# Longest Movie
# ======================================================

longest_movie = max(
    movie_data,
    key=lambda movie: movie["Runtime"]
)

print("\nLongest Movie")
print("-" * 40)

print(
    longest_movie["Title"],
    f"{longest_movie['Runtime']} min"
)

# ======================================================
# Language Count
# ======================================================

language_count = {}

for movie in movie_data:

    languages = movie["Language"].split(", ")

    for lang in languages:

        language_count[lang] = (
            language_count.get(lang, 0) + 1
        )

print("\nLanguage Statistics")
print("-" * 40)

for language, count in language_count.items():

    print(
        f"{language:<15} {count}"
    )

# ======================================================
# Export CSV
# ======================================================

with open(
    "movie_analytics.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=movie_data[0].keys()
    )

    writer.writeheader()

    writer.writerows(movie_data)

print("\nmovie_analytics.csv successfully created!")
```

Q23. API Performance Monitoring System — Using JSONPlaceholder API endpoints (/posts, /users, /comments, /todos), measure response time, payload size, and success rate. Detect slow endpoints, calculate average response time, rank endpoints by performance efficiency, and export monitoring results into api_monitoring_report.json.

```python
import requests
import json
import time

BASE_URL = "https://jsonplaceholder.typicode.com"

endpoints = [
    "/posts",
    "/users",
    "/comments",
    "/todos"
]

monitoring_data = []

successful_requests = 0
total_response_time = 0

for endpoint in endpoints:

    url = BASE_URL + endpoint

    start = time.perf_counter()

    response = requests.get(url)

    end = time.perf_counter()

    response_time = end - start

    payload_size = len(response.content)

    status_code = response.status_code

    success = status_code == 200

    if success:
        successful_requests += 1

    total_response_time += response_time

    performance_score = (
        100
        - response_time * 100
        - payload_size / 1000
    )

    monitoring_data.append({
        "Endpoint": endpoint,
        "Status Code": status_code,
        "Success": success,
        "Response Time (sec)": round(response_time, 4),
        "Payload Size (bytes)": payload_size,
        "Performance Score": round(performance_score, 2)
    })

# ====================================================
# Average Response Time
# ====================================================

average_response = total_response_time / len(endpoints)

print("=" * 60)
print(f"Average Response Time : {average_response:.4f} sec")
print("=" * 60)

# ====================================================
# Success Rate
# ====================================================

success_rate = successful_requests / len(endpoints) * 100

print(f"\nSuccess Rate : {success_rate:.2f}%")

# ====================================================
# Slow Endpoints
# ====================================================

print("\nSlow Endpoints (>0.2 sec)")
print("-" * 40)

found = False

for endpoint in monitoring_data:

    if endpoint["Response Time (sec)"] > 0.2:

        print(
            endpoint["Endpoint"],
            f"{endpoint['Response Time (sec)']} sec"
        )

        found = True

if not found:
    print("No slow endpoints.")

# ====================================================
# Largest Payload
# ====================================================

largest_payload = max(
    monitoring_data,
    key=lambda endpoint: endpoint["Payload Size (bytes)"]
)

print("\nLargest Payload")
print("-" * 40)

print(
    largest_payload["Endpoint"],
    f"{largest_payload['Payload Size (bytes)']} bytes"
)

# ====================================================
# Fastest Endpoint
# ====================================================

fastest = min(
    monitoring_data,
    key=lambda endpoint: endpoint["Response Time (sec)"]
)

print("\nFastest Endpoint")
print("-" * 40)

print(
    fastest["Endpoint"],
    f"{fastest['Response Time (sec)']} sec"
)

# ====================================================
# Slowest Endpoint
# ====================================================

slowest = max(
    monitoring_data,
    key=lambda endpoint: endpoint["Response Time (sec)"]
)

print("\nSlowest Endpoint")
print("-" * 40)

print(
    slowest["Endpoint"],
    f"{slowest['Response Time (sec)']} sec"
)

# ====================================================
# Ranking
# ====================================================

monitoring_data.sort(
    key=lambda endpoint: endpoint["Performance Score"],
    reverse=True
)

print("\nPerformance Ranking")
print("-" * 40)

for index, endpoint in enumerate(monitoring_data, start=1):

    print(
        f"{index}. "
        f"{endpoint['Endpoint']:<12}"
        f"{endpoint['Performance Score']}"
    )

# ====================================================
# Export JSON
# ====================================================

report = {
    "Average Response Time": round(average_response, 4),
    "Success Rate": round(success_rate, 2),
    "Monitoring Results": monitoring_data
}

with open(
    "api_monitoring_report.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        report,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\napi_monitoring_report.json created successfully!")
```

Q25. Insert Data Into SQLite
Insert at least 3 student records into the students table and print all rows.

```python
import sqlite3

# Database bilan bog'lanish
connection = sqlite3.connect("academy.db")

# Cursor yaratish
cursor = connection.cursor()

# 3 ta talaba qo'shish
students = [
    ("Ali Valiyev", 20, "Python"),
    ("Vali Karimov", 22, "Data Science"),
    ("Madina Xasanova", 19, "SQL")
]

cursor.executemany("""
INSERT INTO students (full_name, age, course)
VALUES (?, ?, ?)
""", students)

# O'zgarishlarni saqlash
connection.commit()

# Barcha ma'lumotlarni olish
cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

print("=" * 50)
print("Students Table")
print("=" * 50)

for row in rows:
    print(row)

# Bog'lanishni yopish
connection.close()
```
