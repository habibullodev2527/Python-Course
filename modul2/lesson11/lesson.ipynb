```python
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 
'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

word = input("So'z kiriting: ").lower()
result = ""

for harf in word:
    if harf in alphabet:
        index = alphabet.index(harf)
        new_index = (index + 3) % len(alphabet)
        result += alphabet[new_index]
    else:
        result += harf

print("Natija:", result)
            
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = 'https://asaxiy.uz/uz/product/knigi/badiy-adabiyot/page=3'

response = requests.get(url)
response.status_code
```

```python
html_content = response.text
soup = BeautifulSoup(html_content, "html.parser")
print(soup.prettify())
```

```python
book_name_price_rating = soup.find_all('div', attrs={'class': 'product__item-info position-relative'})
book_name_price_rating
```

## Book Name

```python
book_name = [i.find('span').get_text().strip() for i in book_name_price_rating]
book_name
```

## Price

```python
price = [i.find('span', attrs={'class': 'product__item-price'}).get_text().strip(" so'm") for i in book_name_price_rating]
price
```

## Rating

```python
rating = [i.find('div', attrs={'class': 'product__item-rating--stars'}).get_text() for i in book_name_price_rating]
rating
```
