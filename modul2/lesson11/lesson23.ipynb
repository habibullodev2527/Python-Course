```python
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

soz = input('Soz kiriting').lower().strip()

natija = []
step = int(input('Qadamni kiriting'))

for i in soz:
    index = alphabet.index(i)
    if index + step < len(alphabet):
        natija.append(alphabet[index + step])
    else:
        natija.append(alphabet[step - (len(alphabet) - index)])

('').join(natija)
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

url = 'https://editorial.rottentomatoes.com/guide/best-movies-21st-century/'

response = requests.get(url)
response.status_code
```

```python
html_content = response.text
soup = BeautifulSoup(html_content, 'html.parser')
print(soup.prettify())
```

```python
name_year_scores = soup.find_all('div', attrs={'class':'meta-data-wrapper'})
name_year_scores
```

## Kinoni nomlari

```python
# names = []
# for i in name_year_scores:
#     each_name = i.find('a').get_text()
#     names.append(each_name)

names = [i.find('a').get_text() for i in name_year_scores]
names
```

```python
len(names)
```

## Yillarini ajratib olish

```python
years = [i.find('span', attrs={'class':'meta-year'}).get_text().strip('()') for i in name_year_scores]
years
```

```python
len(years)
```

## Scores ajratib olish

```python
all_scores = [i.find_all('span', attrs={'class':'tMeterScore'}) for i in name_year_scores]
tMeterScore = [i[0].get_text() for i in all_scores]
pMeterScore = [i[1].get_text() for i in all_scores]
```

```python
tMeterScore
```

```python
pMeterScore
```

## Ranks ajratib olish

```python
ranks = [i.get_text().strip('#') for i in soup.find_all('span', attrs={'class':'indicator'})]
ranks
```

```python
len(ranks)
```

```python
movies_info = pd.DataFrame({'Rank':ranks, 'Movie_name':names, 'T_Score': tMeterScore, 
                            'P_Score':pMeterScore, 'Year':years})
movies_info
```

```python
movies_info.to_csv('movies.csv', index=False)
```

```python
import sqlite3

with sqlite3.connect('movies.db') as conn:
    movies_info.to_sql('Movies', conn, if_exists='replace', index=False)
```
