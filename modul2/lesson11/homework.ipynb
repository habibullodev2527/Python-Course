## Homework - 23 Web Scraping 

Using Python web scraping tools, extract the following information from the provided Rotten Tomatoes page:

- Cartoon/Movie Name
- Release Year
- Tomatometer Score
- Rank

Requirements:
1. Scrape all available records from the page.
2. Store the extracted data in a Pandas DataFrame.
3. Create a SQLite database named 'Cartoon.db'.
4. Save the DataFrame into a table named 'Cartoons'.
5. Verify that the data has been successfully inserted into the database.

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup


url = "https://editorial.rottentomatoes.com/guide/movies-100-percent-score-rotten-tomatoes/"

response = requests.get(url)
response.status_code
```

### 1. Scrape all available records from the page.

```python
html_content = response.text
soup = BeautifulSoup(html_content, "html.parser")
print(soup.prettify())
```

```python
name_year_scores = soup.find_all('div', attrs={'class':'meta-data-wrapper'})
name_year_scores
```

### Names

```python
names = [i.find('a').get_text() for i in name_year_scores]
names
```

### Years

```python
years = [i.find('span', attrs={'class':'meta-year'}).get_text().strip('()') for i in name_year_scores]
years
```

### Scores

```python
scores = [i.find('span', attrs={'class': 'tMeterScore'}).get_text() for i in name_year_scores]
scores
```

### Ranks

```python
ranks = [i.get_text().strip('#') for i in soup.find_all('span', attrs={'class': 'indicator'})]
ranks
```

### 2. Store the extracted data in a Pandas DataFrame.

```python
cartoons = pd.DataFrame({'MovieName': names, 'ReleaseYear': years, 'TomatometerScore': scores, 'Rank': ranks})

cartoons
```

### 3. Create a SQLite database named 'Cartoon.db'.

### 4. Save the DataFrame into a table named 'Cartoons'.

```python
import sqlite3

with sqlite3.connect('Cartoon.db') as conn:
    cartoons.to_sql('Cartoons', conn, if_exists='replace', index=False)
```

### 5. Verify that the data has been successfully inserted into the database.

```python
conn = sqlite3.connect("Cartoon.db")

result = pd.read_sql("SELECT * FROM Cartoons", conn)

result

conn.close()
```
