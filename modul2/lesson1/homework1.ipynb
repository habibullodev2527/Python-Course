Q2. Age Calculator
Ask the user to enter their birthdate. Calculate and print their age in years, months, and days.

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

birth = input("Tug'ilgan kuningizni kiriting (YYYY-MM-DD): ")

birthdate = datetime.strptime(birth, "%Y-%m-%d")

today = datetime.today()

age = relativedelta(today, birthdate)

print(f"Age: {age.years} years, {age.months} months, {age.days} days")
```

Q4. Days Until Next Birthday
Calculate and print the number of days remaining until the user's next birthday.

```python
from datetime import datetime

birth = input("Enter your birthdate (YYYY-MM-DD): ")

birthdate = datetime.strptime(birth, "%Y-%m-%d").date()
today = datetime.today().date()

# Joriy yil uchun tug'ilgan kun
next_birthday = birthdate.replace(year=today.year)

# Agar tug'ilgan kun o'tib ketgan bo'lsa, keyingi yilni olamiz
if next_birthday < today:
    next_birthday = next_birthday.replace(year=today.year + 1)

days_left = (next_birthday - today).days

print(f"Days until next birthday: {days_left}")
```

Q10. Year Checker
Check whether the current year is a leap year.

```python
from datetime import datetime

year = datetime.now().year

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

Q14. Time Delta
Calculate the difference in seconds between two timestamps.

```python
from datetime import datetime

time1 = input("Enter first timestamp (YYYY-MM-DD HH:MM:SS): ")
time2 = input("Enter second timestamp (YYYY-MM-DD HH:MM:SS): ")

t1 = datetime.strptime(time1, "%Y-%m-%d %H:%M:%S")
t2 = datetime.strptime(time2, "%Y-%m-%d %H:%M:%S")

difference = abs((t2 - t1).total_seconds())

print(f"Difference: {int(difference)} seconds")
```

Q17. Password Strength Checker
Validate password strength using multiple conditions.

```python
import re

password = input("Enter password: ")

if (
    len(password) >= 8
    and re.search(r"[A-Z]", password)
    and re.search(r"[a-z]", password)
    and re.search(r"[0-9]", password)
    and re.search(r"[!@#$%^&*_-]", password)
):
    print("Strong Password")
else:
    print("Weak Password")
```

Q18. Workdays Only
Calculate working days between two dates.

```python
from datetime import datetime, timedelta

start_date = input("Start date: ")
end_date = input("End date: ")

start = datetime.strptime(start_date, "%Y-%m-%d")
end = datetime.strptime(end_date, "%Y-%m-%d")

count = 0

current = start

while current <= end:

    if current.weekday() < 5:
        count += 1

    current += timedelta(days=1)

print(count)
```

Q20. Timezone Converter
Convert datetime values between timezones.

```python
from datetime import datetime
from zoneinfo import ZoneInfo

date = input("Enter datetime (YYYY-MM-DD HH:MM:SS): ")

# Kiritilgan vaqtni UTC deb olamiz
utc_time = datetime.strptime(date, "%Y-%m-%d %H:%M:%S")
utc_time = utc_time.replace(tzinfo=ZoneInfo("UTC"))

# Toshkent vaqtiga o'tkazish
tashkent_time = utc_time.astimezone(ZoneInfo("Asia/Tashkent"))

print("UTC Time      :", utc_time)
print("Tashkent Time :", tashkent_time)
```

Q21. Log Parser
Identify valid emails and dates and detect invalid dates using datetime validation.


```python
import re
from datetime import datetime

raw_data = """
1. Ali, email: ali@gmail.com, date: 2024-05-10
2. Vali, email: vali@mail.ru, date: 2023-02-30
3. Sara, email: sara_85@univ.uz, date: 2022-12-15
4. Test, email: xato-email@, date: 2024-13-01
"""

# Emaillarni topish
emails = re.findall(
    r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",
    raw_data
)

print("Valid Emails:")
for email in emails:
    print(email)

print("\nDates:")

# Sanalarni topish
dates = re.findall(r"\d{4}-\d{2}-\d{2}", raw_data)

for date in dates:
    try:
        datetime.strptime(date, "%Y-%m-%d")
        print(f"{date} -> Valid")
    except ValueError:
        print(f"{date} -> Invalid")
```

Q22. Countdown Timer
Implement a countdown timer updating every second.

```python
import time

seconds = int(input("Enter countdown time (seconds): "))

while seconds > 0:
    print(seconds)
    time.sleep(1)
    seconds -= 1

print("Time's up!")
```

Q23. Text Anonymizer
Replace all emails and phone numbers with [HIDDEN].

```python
import re

raw_data = """
Ali: ali@gmail.com, Phone: +998901234567
Vali: vali@mail.ru, Phone: 998 (93) 456-78-90
Sara: sara_85@univ.uz, Phone: 712345678
"""

# Email pattern
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

# Phone pattern
phone_pattern = r"\+?\d[\d\s()-]*"

# Email va telefonlarni yashirish
text = re.sub(email_pattern, "[HIDDEN]", raw_data)
text = re.sub(phone_pattern, "[HIDDEN]", text)

print(text)
```
