```python
create_table = """
    create table People(
        FirstName text,
        LastName text,
        Age int
    );
"""

insert_table = """
    insert into People values (
        'John',
        'Doe',
        25
    );
"""

connection = sqlite3.connect("lesson15.db")
cursor = connection.cursor()
cursor.execute(create_table)
cursor.execute(insert_table)
connection.commit()     # barcha o'zgarishlarni database ichiga saxranit qiladi
connection.close()
```

```python
create_table = """
    create table student(
        FirstName text,
        LastName text,
        Age int
    );
"""

insert_table = """
    insert into student values 
    ('John', 'Doe', 25), 
    ('Ali', 'Valiyev', 30);
"""

with sqlite3.connect('lesson15.db') as conn:
    cursor = conn.cursor()
    cursor.execute(create_table)
    cursor.execute(insert_table)
```

* executescript() - bir vaqtni o'zida bir nechta buyruqlarni ishlatish uchun

```python
sql_query = """
Drop table if exists people;
create table people(fullname text, age int, score integer);
insert into people values 
('Anvar Rahmatullayev', 18, 90), 
('Malika Abdullayeva', 22, 88);
"""

with sqlite3.connect('test.db') as connection:
    cursor = connection.cursor()
    cursor.executescript(sql_query)
```

* executemany() - parameter asosida malumot kiritish

```python
people_values = (
    ("Abbos Fayzullayev", 21, 100),
    ('Eldor Shomurodov', 34, 98),
    ('Abduqodir Husanov', 22, 99)
)

with sqlite3.connect('test.db') as connection:
    cursor = connection.cursor()
    cursor.executemany("insert into people values (?, ?, ?)", people_values)
```

```python
fullname = input("Ism familyangizni kiriting: ")
age = int(input("Yoshingizni kiriting: "))
score = int(input("Olgan ballingizni kiriting: "))

query = (
    "insert into People values"
    f"('{fullname}', {age}, {score})"
)

with sqlite3.connect('test.db') as connection:
    cursor = connection.cursor()
    cursor.execute(query)
```

### working with other SQL databases

```python
import pyodbc

connection = pyodbc.connect(
    'DRIVER={SQL Server};'
    'Server=PC;'
    'Database=TSQL2012;'
    'trusted_Connection=yes;'  
)

print('Connected successfully')

cursor = connection.cursor()

cursor.execute("""
                Update [Production].[Products] 
                set productname = 'Product10' 
                where productid = 10;""")
result = cursor.execute("select * from [Production].[Products];")
result.fetchall()
```
