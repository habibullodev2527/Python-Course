```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/users"

response = requests.get(url)

if response.status_code == 200:
    data = response.json()

    result = {}

    for user in data:
        if user['id'] == 3:
            result = {
                "name": user["name"],
                "email": user["email"],
                "address": {
                    "city": user["address"]["city"]
                }
            }

    json_data = json.dumps(result)
    print(json_data)
    
else:
    print("Xatolik", response.status_code)
```
