Q21. Profitability Intelligence Dashboard — Using corporate_data.db, combine employees and sales tables. Calculate Profit = Revenue - Discount_Applied, calculate total profit for each employee, identify employees with negative profit, rank departments by total revenue, calculate department efficiency score, and export department_profitability.csv.

```python
import sqlite3
import csv

# Database bilan ulanish
connection = sqlite3.connect("corporate_data.db")
cursor = connection.cursor()

# ==========================================================
# Q1. Employee Profitability
# ==========================================================

print("=" * 60)
print("TOTAL PROFIT FOR EACH EMPLOYEE")
print("=" * 60)

cursor.execute("""
    SELECT
        e.Employee_ID,
        e.Full_Name,
        SUM(s.Revenue - s.Discount_Applied) AS TotalProfit
    FROM employees e
    JOIN sales s
        ON e.Employee_ID = s.Employee_ID
    GROUP BY
        e.Employee_ID,
        e.Full_Name
    ORDER BY TotalProfit DESC;
""")

employee_profit = cursor.fetchall()

for row in employee_profit:
    print(row)

# ==========================================================
# Q2. Employees with Negative Profit
# ==========================================================

print("\n")
print("=" * 60)
print("EMPLOYEES WITH NEGATIVE PROFIT")
print("=" * 60)

cursor.execute("""
    SELECT
        e.Employee_ID,
        e.Full_Name,
        SUM(s.Revenue - s.Discount_Applied) AS TotalProfit
    FROM employees e
    JOIN sales s
        ON e.Employee_ID = s.Employee_ID
    GROUP BY
        e.Employee_ID,
        e.Full_Name
    HAVING SUM(s.Revenue - s.Discount_Applied) < 0;
""")

negative_profit = cursor.fetchall()

for row in negative_profit:
    print(row)

# ==========================================================
# Q3. Department Revenue Ranking
# ==========================================================

print("\n")
print("=" * 60)
print("DEPARTMENT REVENUE RANKING")
print("=" * 60)

cursor.execute("""
    SELECT
        e.Department,
        SUM(s.Revenue) AS TotalRevenue
    FROM employees e
    JOIN sales s
        ON e.Employee_ID = s.Employee_ID
    GROUP BY e.Department
    ORDER BY TotalRevenue DESC;
""")

department_revenue = cursor.fetchall()

for row in department_revenue:
    print(row)

# ==========================================================
# Q4. Department Efficiency Score
# ==========================================================

print("\n")
print("=" * 60)
print("DEPARTMENT EFFICIENCY SCORE")
print("=" * 60)

cursor.execute("""
    SELECT
        e.Department,
        SUM(s.Revenue) AS TotalRevenue,
        SUM(s.Revenue - s.Discount_Applied) AS TotalProfit,
        ROUND(
            (SUM(s.Revenue - s.Discount_Applied) * 100.0)
            / SUM(s.Revenue),
            2
        ) AS EfficiencyScore
    FROM employees e
    JOIN sales s
        ON e.Employee_ID = s.Employee_ID
    GROUP BY e.Department
    ORDER BY EfficiencyScore DESC;
""")

department_efficiency = cursor.fetchall()

for row in department_efficiency:
    print(row)

# ==========================================================
# Q5. Export CSV
# ==========================================================

with open(
    "department_profitability.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    writer.writerow([
        "Department",
        "TotalRevenue",
        "TotalProfit",
        "EfficiencyScore"
    ])

    writer.writerows(department_efficiency)

print("\nCSV file created successfully!")

connection.close()
```

Q22. Revenue Fraud & Anomaly Detection — Using sales table, identify suspicious transactions where Discount_Applied > 25, Revenue > 2x department average, or cancelled sales still contain revenue. Calculate anomaly percentage by department, rank suspicious transactions by risk score, and export suspicious_sales.csv.

```python
import sqlite3
import csv

connection = sqlite3.connect("corporate_data.db")
cursor = connection.cursor()

# ==========================================================
# Q22. Revenue Fraud & Anomaly Detection
# ==========================================================

# 1. Department Average Revenue
cursor.execute("""
CREATE TEMP VIEW IF NOT EXISTS DepartmentAverage AS
SELECT
    e.Department,
    AVG(s.Revenue) AS AvgRevenue
FROM employees e
JOIN sales s
    ON e.Employee_ID = s.Employee_ID
GROUP BY e.Department;
""")

# 2. Suspicious Transactions
cursor.execute("""
SELECT
    s.Sale_ID,
    e.Employee_ID,
    e.Full_Name,
    e.Department,
    s.Revenue,
    s.Discount_Applied,
    s.Status,

    CASE
        WHEN s.Discount_Applied > 25 THEN 1
        ELSE 0
    END +

    CASE
        WHEN s.Revenue > 2 * d.AvgRevenue THEN 2
        ELSE 0
    END +

    CASE
        WHEN s.Status = 'Cancelled'
             AND s.Revenue > 0 THEN 3
        ELSE 0
    END AS RiskScore

FROM sales s

JOIN employees e
    ON s.Employee_ID = e.Employee_ID

JOIN DepartmentAverage d
    ON e.Department = d.Department

WHERE
      s.Discount_Applied > 25
   OR s.Revenue > 2 * d.AvgRevenue
   OR (
        s.Status = 'Cancelled'
        AND s.Revenue > 0
      )

ORDER BY RiskScore DESC, s.Revenue DESC;
""")

suspicious_sales = cursor.fetchall()

print("=" * 80)
print("SUSPICIOUS TRANSACTIONS")
print("=" * 80)

for row in suspicious_sales:
    print(row)

# ==========================================================
# Department Anomaly Percentage
# ==========================================================

cursor.execute("""
SELECT
    e.Department,

    COUNT(
        CASE
            WHEN
                 s.Discount_Applied > 25
              OR s.Revenue > 2 * d.AvgRevenue
              OR (
                    s.Status='Cancelled'
                    AND s.Revenue>0
                 )
            THEN 1
        END
    ) AS SuspiciousSales,

    COUNT(*) AS TotalSales,

    ROUND(
        COUNT(
            CASE
                WHEN
                     s.Discount_Applied > 25
                  OR s.Revenue > 2 * d.AvgRevenue
                  OR (
                        s.Status='Cancelled'
                        AND s.Revenue>0
                     )
                THEN 1
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS AnomalyPercentage

FROM employees e

JOIN sales s
    ON e.Employee_ID = s.Employee_ID

JOIN DepartmentAverage d
    ON e.Department = d.Department

GROUP BY e.Department

ORDER BY AnomalyPercentage DESC;
""")

department_result = cursor.fetchall()

print("\n")
print("=" * 80)
print("ANOMALY PERCENTAGE BY DEPARTMENT")
print("=" * 80)

for row in department_result:
    print(row)

# ==========================================================
# Export suspicious_sales.csv
# ==========================================================

with open(
    "suspicious_sales.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    writer.writerow([
        "Sale_ID",
        "Employee_ID",
        "Full_Name",
        "Department",
        "Revenue",
        "Discount_Applied",
        "Status",
        "RiskScore"
    ])

    writer.writerows(suspicious_sales)

print("\nCSV file created successfully!")

connection.close()
```

Q23. Executive Department Performance System — Using employees and sales tables, calculate department revenue, average salary, employee performance score, and revenue-per-salary efficiency. Identify departments with high salary but low performance, detect top employees in each department, rank departments by efficiency, and export executive_dashboard.csv.

```python
import sqlite3
import csv

connection = sqlite3.connect("corporate_data.db")
cursor = connection.cursor()

# ==========================================================
# Q23. Executive Department Performance System
# ==========================================================

# ----------------------------------------------------------
# Part 1-4
# Department Revenue
# Average Salary
# Average Performance
# Revenue per Salary Efficiency
# ----------------------------------------------------------

cursor.execute("""
SELECT
    e.Department,
    ROUND(SUM(s.Revenue),2) AS TotalRevenue,
    ROUND(AVG(e.Salary),2) AS AvgSalary,
    ROUND(AVG(e.Performance_Score),2) AS AvgPerformance,

    ROUND(
        SUM(s.Revenue) / AVG(e.Salary),
        2
    ) AS RevenuePerSalary

FROM employees e

JOIN sales s
ON e.Employee_ID = s.Employee_ID

GROUP BY e.Department

ORDER BY RevenuePerSalary DESC;
""")

department_dashboard = cursor.fetchall()

print("="*80)
print("DEPARTMENT DASHBOARD")
print("="*80)

for row in department_dashboard:
    print(row)

# ----------------------------------------------------------
# Part 5
# High Salary but Low Performance
# ----------------------------------------------------------

cursor.execute("""
SELECT
    e.Department,
    ROUND(AVG(e.Salary),2) AS AvgSalary,
    ROUND(AVG(e.Performance_Score),2) AS AvgPerformance

FROM employees e

GROUP BY e.Department

HAVING
    AVG(e.Salary) > (
        SELECT AVG(Salary)
        FROM employees
    )

AND

    AVG(e.Performance_Score) < (
        SELECT AVG(Performance_Score)
        FROM employees
    );
""")

high_salary_low_perf = cursor.fetchall()

print("\n")
print("="*80)
print("HIGH SALARY BUT LOW PERFORMANCE")
print("="*80)

for row in high_salary_low_perf:
    print(row)

# ----------------------------------------------------------
# Part 6
# Top Employee in Each Department
# ----------------------------------------------------------

cursor.execute("""
SELECT
    Department,
    Full_Name,
    Performance_Score

FROM (

    SELECT
        Department,
        Full_Name,
        Performance_Score,

        ROW_NUMBER() OVER(
            PARTITION BY Department
            ORDER BY Performance_Score DESC
        ) AS rn

    FROM employees

)

WHERE rn = 1;
""")

top_employees = cursor.fetchall()

print("\n")
print("="*80)
print("TOP EMPLOYEES")
print("="*80)

for row in top_employees:
    print(row)

# ----------------------------------------------------------
# Part 7
# Department Efficiency Ranking
# ----------------------------------------------------------

cursor.execute("""
SELECT
    e.Department,

    ROUND(
        SUM(s.Revenue) / AVG(e.Salary),
        2
    ) AS Efficiency

FROM employees e

JOIN sales s
ON e.Employee_ID = s.Employee_ID

GROUP BY e.Department

ORDER BY Efficiency DESC;
""")

department_rank = cursor.fetchall()

print("\n")
print("="*80)
print("DEPARTMENT EFFICIENCY RANKING")
print("="*80)

for row in department_rank:
    print(row)

# ----------------------------------------------------------
# Part 8
# Export CSV
# ----------------------------------------------------------

with open(
    "executive_dashboard.csv",
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.writer(file)

    writer.writerow([
        "Department",
        "TotalRevenue",
        "AverageSalary",
        "AveragePerformance",
        "RevenuePerSalary"
    ])

    writer.writerows(department_dashboard)

print("\nexecutive_dashboard.csv created successfully!")

connection.close()
```
