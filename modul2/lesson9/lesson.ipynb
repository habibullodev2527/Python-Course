```python
import pandas as pd

data = pd.read_csv('ibm.csv')

data
```

```python
def dfValueStatus(series):
    if series < 50:
        return "Juda past narx"
    elif 50 <= series <= 100:
        return "O'rtacha narx"
    else:
        return "Yuqori narx"
    
data["Status"] = data["Close"].apply(dfValueStatus)

data
```
