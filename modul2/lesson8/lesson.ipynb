```python
import pandas as pd

data = {
    "Student": ["Ali", "Vali", "Sami", "Lola", "Nodir",
                "Aziza", "Jasur", "Malika", "Timur", "Zarina"],

    "Group": ["A", "B", "A", "B", "A",
              "B", "A", "B", "A", "B"],

    "Subject": ["Math", "Math", "English", "English", "Math",
                "English", "Math", "Math", "English", "English"],

    "Score": [85, 67, 90, 74, 58,
              88, 95, 60, 77, 82],

    "Attendance": [90, 75, 95, 80, 60,
                   92, 98, 70, 85, 88]
}

data = pd.DataFrame(data)

data 
```

```python
data[data['Score'] > 80]
```

```python
data[['Student', 'Subject', 'Score']]
```

```python
data.groupby('Group', as_index=False).agg(AverageScore=('Score', 'mean'))
```

```python
result = (
    data.groupby('Subject', as_index=False)
    .agg(
        MaxScore=('Score', 'max'),
        AverageAttendance=('Attendance', 'mean')
    )
)

result
```

```python
result = (
    data[data['Attendance'] > 80].groupby('Group', as_index=False)
    .agg(StudentCount=('Student', 'count'))
)

result
```

```python
from sqlalchemy import create_engine

engine = create_engine(r'sqlite:///data\chinook.db')
```

```python
genres = pd.read_sql_table('genres', engine)

tracks = pd.read_sql_table('tracks', engine)

genres
```

```python
genres.merge(tracks, on='GenreId').groupby('Name_x').agg({"TrackId":'count'})\
    .rename(columns={'TrackId':'Number of Tracks'}).sort_values(by='Number of Tracks', ascending=False).head()
```
