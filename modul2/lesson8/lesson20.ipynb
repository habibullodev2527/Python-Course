```python
import pandas as pd

data = {
    "Student": ["Ali", "Vali", "Sami", "Lola", "Nodir",
                "Aziza", "Jasur", "Malika", "Timur", "Zarina"],

    "Group": ["A", "B", "A", "B", "A",
              "B", "A", "B", "A", "B"],

    "Subject": ["Math", "Math", "English", "English", "Math",
                "English", "Math", "Math", "English", "English"],

    "Score": [85, 67, 90, 74, 58,
              88, 95, 60, 77, 82],

    "Attendance": [90, 75, 95, 80, 60,
                   92, 98, 70, 85, 88]
}

df = pd.DataFrame(data)
df
```

```python
df[df['Score'] > 80]
```

```python
df[['Student', 'Subject', 'Score']]
```

```python
df.groupby('Group').agg(Avg_score = ('Score','mean'))
```

```python
df.groupby('Subject').agg({'Score': 'max', 'Attendance':'mean'})
```

```python
df[df['Attendance'].gt(80)].groupby('Group')['Student'].count()
```

```python
df[df['Attendance'].gt(80)].groupby('Group').agg(student_count = ('Student', 'count'))
```
