```python
weight_loss = pd.read_csv('data/weight_loss.csv')
weight_loss.head()
```

```python
weight_loss.tail()
```

```python
weight_loss.groupby('Name').agg({"Weight":['first', 'last']})
```

```python
weight_loss.groupby('Name').agg({'Weight':'sqrt_mean'})
```

```python
def sqrt_mean(series):
    positive_values = [x ** 0.5 for x in series if x >= 0]
    return sum(positive_values) / len(positive_values) if positive_values else None
```

```python
weight_loss.groupby('Name').agg({'Weight': sqrt_mean})
```

```python
sqrt_mean([64])
```

```python
sales_data = pd.read_csv('data/sales_data.csv')
sales_data
```

You are given a dataset containing sales data for an e-commerce website. The dataset (`task\sales_data.csv`) has the following columns:

- `Date`: Date of the sale.
- `Product`: Name of the product sold.
- `Category`: Category to which the product belongs.
- `Quantity`: Number of units sold.
- `Price`: Price per unit.

**Tasks:**

1. Group the data by the `Category` column and calculate the following aggregate statistics for each category:
   - Total quantity sold.
   - Average price per unit.

```python
sales_data.groupby('Category').agg({'Quantity':'sum', 'Price':'mean'}).round(1)
```

```python
sales_data['Total_Sales'] = sales_data['Quantity'] * sales_data['Price']
```

```python
sales_data
```

```python
sales_data.iloc[sales_data.groupby('Category')['Quantity'].idxmax()]    # har bir category boyicha eng maximum sotilgan mahsulotlar soni
```

```python
sales_data.iloc[sales_data.groupby('Date')['Total_Sales'].idxmax()]['Date'].size
```
