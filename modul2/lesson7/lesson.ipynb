```python
import pandas as pd

flights = pd.read_csv('data/flights.csv')

flights
```

```python
flights.groupby(['AIRLINE', 'WEEKDAY']).agg({
    'CANCELLED': ['min', 'max', 'count'],
    'DIVERTED': ['mean', 'max', 'sum']
})
```

```python
sales_data = pd.read_csv('data/sales_data.csv')

sales_data
```

```python
sales_data.groupby('Category').agg({
    'Quantity': 'sum',
    'Price': 'mean'
})
```
