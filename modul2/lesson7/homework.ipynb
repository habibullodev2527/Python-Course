Q21. Gender Salary Gap Intelligence Report
Using the population table from population.db, create a salary gap report by state and gender.

Your final report should include:

-	state
-	MaleAvgSalary
-	FemaleAvgSalary
-	OverallAvgSalary
-	SalaryGap
-	EqualityIndex
-	InequalityRank

Use these formulas:

SalaryGap = abs(MaleAvgSalary - FemaleAvgSalary)

EqualityIndex = 1 - (SalaryGap / OverallAvgSalary)

Requirements:

1.	Group by state and gender.
2.	Calculate average salary.
3.	Reshape the result using pivot_table() or unstack().
4.	Calculate SalaryGap.
5.	Calculate EqualityIndex.
6.	Rank states by highest salary gap.
7.	Export the final result as:

salary_gap_report.csv

Note:

If the dataset contains gender values other than Male/Female, handle them carefully and explain your approach in a comment.

```python
conn = sqlite3.connect('data/population.db')

population = pd.read_sql_query("SELECT * FROM population", conn)

# Group by state and gender. Calculate average salary.

result = (
    population.groupby(['state', 'gender'], as_index=False)
    .agg(AverageSalary=('salary', 'mean'))   
)

# Reshape the result using pivot_table() or unstack().

result = result.pivot_table(
    index='state',
    columns='gender',
    values='AverageSalary'
).reset_index()

# rename columns name
result = result.rename(columns={
    'Male': 'MaleAvgSalary',
    'Female': 'FemaleAvgSalary'
})

# Overall average salary
overall_avg = (
    population
    .groupby('state', as_index=False)
    .agg(OverallAvgSalary=('salary', 'mean'))
)

# Add OverallAvgSalary
result = result.merge(overall_avg, on='state')

# Calculate SalaryGap.

result['SalaryGap'] = abs(
    result['MaleAvgSalary'] - result['FemaleAvgSalary']
)

# Calculate EqualityIndex.

result['EqualityIndex'] = (
    1 - result['SalaryGap'] / result['OverallAvgSalary']
).round(4)

# 6.	Rank states by highest salary gap.

result = result.sort_values(by='SalaryGap', ascending=False)

# 7.	Export the final result as: salary_gap_report.csv

result.to_csv('data/salary_gap_report.csv', index=False)

result

conn.close()
```

Q22. Revenue Performance Analytics Report
Using sales_data.csv, create a revenue performance report.
First create:
Revenue = Quantity * Price
Your category-level report should include:
•	Category
•	TotalRevenue
•	TotalQuantity
•	AveragePrice
•	ProductCount
•	PerformanceStatus
Rules:
If TotalRevenue >= average category revenue:
    PerformanceStatus = "Above Average"
else:
    PerformanceStatus = "Underperforming"
Also create a product-level report with:
•	Product
•	Category
•	ProductRevenue
•	ProductQuantity
Requirements:
1.	Calculate total revenue by category.
2.	Calculate total quantity by category.
3.	Calculate average price by category.
4.	Count unique products by category.
5.	Identify underperforming categories.
6.	Find top 10 products by revenue.
7.	Export category report as:
category_performance.csv
8.	Export top product report as:
top_products_by_revenue.csv

```python
# Create Revenue column
sales_data['Revenue'] = sales_data['Quantity'] * sales_data['Price']

# Category-level report
category_report = (
    sales_data
    .groupby('Category', as_index=False)
    .agg(
        TotalRevenue=('Revenue', 'sum'),
        TotalQuantity=('Quantity', 'sum'),
        AveragePrice=('Price', 'mean'),
        ProductCount=('Product', 'nunique')
    )
)

# Calculate average category revenue
average_category_revenue = category_report['TotalRevenue'].mean()

# Create PerformanceStatus column
category_report['PerformanceStatus'] = category_report['TotalRevenue'].apply(
    lambda x: 'Above Average'
    if x >= average_category_revenue
    else 'Underperforming'
)

# Product-level report
product_report = (
    sales_data
    .groupby(['Product', 'Category'], as_index=False)
    .agg(
        ProductRevenue=('Revenue', 'sum'),
        ProductQuantity=('Quantity', 'sum')
    )
)

# Find top 10 products by revenue
top_products = (
    product_report
    .sort_values(by='ProductRevenue', ascending=False)
    .head(10)
)

# Export reports
category_report.to_csv(
    'data/category_performance.csv',
    index=False
)

top_products.to_csv(
    'data/top_products_by_revenue.csv',
    index=False
)

# Display results
print(category_report)
print(top_products)
```

Requirements:

```python
import pandas as pd

# Load dataset
customer_orders = pd.read_csv('data/customer_orders.csv')

# Create Total_Spent column
customer_orders['Total_Spent'] = (
    customer_orders['Quantity'] * customer_orders['Price']
)

# Customer-level report
customer_report = (
    customer_orders
    .groupby('CustomerID', as_index=False)
    .agg(
        TotalSpent=('Total_Spent', 'sum'),
        TotalOrders=('CustomerID', 'count'),
        TotalQuantity=('Quantity', 'sum'),
        UniqueProductTypes=('Product', 'nunique')
    )
)

# Spending quantiles
q1 = customer_report['TotalSpent'].quantile(0.25)
q3 = customer_report['TotalSpent'].quantile(0.75)

# Customer segmentation
customer_report['CustomerSegment'] = customer_report['TotalSpent'].apply(
    lambda x: (
        'VIP'
        if x >= q3
        else 'Regular'
        if x >= q1
        else 'Low Spender'
    )
)

# Repeat buyer flag
customer_report['IsRepeatBuyer'] = (
    customer_report['UniqueProductTypes'] > 5
)

# Export report
customer_report.to_csv(
    'data/customer_segments.csv',
    index=False
)

# Display result
print(customer_report)
```

Q24. merge() Introduction
Take the first 5 rows from customer_orders.csv.
Create a new small DataFrame manually:
CustomerID, CustomerName, City
Make sure some CustomerID values match the customer_orders.csv data.
Then merge both DataFrames using:
pd.merge()
Merge key:
CustomerID
Display:
•	OrderID
•	CustomerID

•	City

```python
import pandas as pd

# Load dataset
customer_orders = pd.read_csv('data/customer_orders.csv')

# Take first 5 rows
orders = customer_orders.head(5)

# Create a small DataFrame manually
customers = pd.DataFrame({
    'CustomerID': [101, 102, 103, 104, 105],   # Use IDs that exist in your dataset
    'CustomerName': ['Ali', 'Vali', 'Hasan', 'Husan', 'Dilshod'],
    'City': ['Tashkent', 'Samarkand', 'Bukhara', 'Andijan', 'Namangan']
})

# Merge both DataFrames
result = pd.merge(
    orders,
    customers,
    on='CustomerID',
    how='inner'
)

# Display required columns
result = result[
    [
        'OrderID',
        'CustomerID',
        'CustomerName',
        'City',
        'Product',
        'Quantity'
    ]
]

print(result)
```

Q25. concat() Introduction
Create two small DataFrames manually with the same columns:
StudentID, Name, Score
Then vertically combine them using:
pd.concat()
Requirements:
1.	Use ignore_index=True.
2.	Display the final combined DataFrame.
3.	Explain in a comment why ignore_index=True is useful.

```python
import pandas as pd

# Create first DataFrame
students1 = pd.DataFrame({
    'StudentID': [1, 2, 3],
    'Name': ['Ali', 'Vali', 'Hasan'],
    'Score': [85, 90, 78]
})

# Create second DataFrame
students2 = pd.DataFrame({
    'StudentID': [4, 5, 6],
    'Name': ['Husan', 'Dilshod', 'Jasur'],
    'Score': [88, 92, 81]
})

# Combine both DataFrames vertically
# ignore_index=True creates a new continuous index (0, 1, 2, ...)
# instead of keeping the original indexes from each DataFrame.
result = pd.concat(
    [students1, students2],
    ignore_index=True
)

# Display the final DataFrame
print(result)
```
