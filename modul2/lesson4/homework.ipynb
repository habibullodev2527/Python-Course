Q14. Min and Max Search
Create a 10x10 random array and find:
- minimum value
- maximum value
- indices of both values

```python
import numpy as np

arr = np.random.randint(1, 101, (10, 10))

print(arr)

print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

min_index = np.unravel_index(np.argmin(arr), arr.shape)
max_index = np.unravel_index(np.argmax(arr), arr.shape)

print("Minimum index:", min_index)
print("Maximum index:", max_index)
```

Q22. Matrix Reshape and Analysis

Create an array containing numbers from 1 to 100.

Perform the following:
- reshape it into a 10x10 matrix
- print the first row
- print the last column
- print all even numbers
- find the maximum and minimum values
- calculate the average value

```python
import numpy as np

# 1 dan 100 gacha bo'lgan sonlardan massiv yaratish
arr = np.arange(1, 101)

# 10x10 matritsaga o'zgartirish
matrix = arr.reshape((10, 10))

# Birinchi qator
first_row = matrix[0]

# Oxirgi ustun
last_column = matrix[:, -1]

# Barcha juft sonlar
even_numbers = matrix[matrix % 2 == 0]

# Maksimum va minimum qiymatlar
maximum = matrix.max()
minimum = matrix.min()

# O'rtacha qiymat
average = matrix.mean()

# Natijalarni chiqarish
print("10x10 Matrix:")
print(matrix)

print("\nFirst row:")
print(first_row)

print("\nLast column:")
print(last_column)

print("\nEven numbers:")
print(even_numbers)

print("\nMaximum value:", maximum)
print("Minimum value:", minimum)

print("\nAverage value:", average)
```

Q23. Matrix Comparison Challenge
Create two 5x5 random matrices.
Perform:
- element-wise addition
- element-wise subtraction
- element-wise multiplication
- determine which matrix has the larger total sum

```python
import numpy as np

# Ikkita 5x5 tasodifiy matritsa yaratish
arr1 = np.random.randint(1, 101, (5, 5))
arr2 = np.random.randint(1, 101, (5, 5))

print("Matrix 1:")
print(arr1)

print("\nMatrix 2:")
print(arr2)

# Element-wise addition
addition = arr1 + arr2

# Element-wise subtraction
subtraction = arr1 - arr2

# Element-wise multiplication
multiplication = arr1 * arr2

print("\nElement-wise Addition:")
print(addition)

print("\nElement-wise Subtraction:")
print(subtraction)

print("\nElement-wise Multiplication:")
print(multiplication)

# Har bir matritsaning umumiy yig'indisi
sum1 = arr1.sum()
sum2 = arr2.sum()

print("\nTotal Sum of Matrix 1:", sum1)
print("Total Sum of Matrix 2:", sum2)

# Qaysi matritsaning yig'indisi katta ekanligini aniqlash
if sum1 > sum2:
    print("\nMatrix 1 has the larger total sum.")
elif sum2 > sum1:
    print("\nMatrix 2 has the larger total sum.")
else:
    print("\nBoth matrices have the same total sum.")
```
